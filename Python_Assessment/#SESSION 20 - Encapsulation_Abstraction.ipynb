```python
#20:1).Create a Python class called Product with a private attribute _price. Initialize _price in the constructor and write a method to display its value.

class Product:
    def __init__(self, price):
        self._price = price  


    def display_price(self):
        print(f"Product Price: ₹{self._price}")


product = Product(558)

product.display_price()
```

```python
#20:2).Add getter and setter methods for the _price attribute in your Product class to safely access and update the price. Make sure the setter prevents setting a negative price.<br><br><em><strong>Hint:</strong> Raise a ValueError if the new price is less than zero.</em>

class Product:
    def __init__(self, price):
        self._price = price  

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative.")
        self._price = new_price


product = Product(667)

print("Current Price: ₹", product.get_price())

product.set_price(1200)
print("Updated Price: ₹", product.get_price())


try:
    product.set_price(-500)
except ValueError as e:
    print("Error:", e)
```

```python
#20:3).Build a class called Playlist that has a private attribute _songs (a list of song names). Write methods to add a song, remove a song, and get the current list of songs using proper encapsulation.

class Playlist:
    def __init__(self):
        self._songs = []  

    
    def add_song(self, song):
        self._songs.append(song)
        print(f"'{song}' added to the playlist.")

    
    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
            print(f"'{song}' removed from the playlist.")
        else:
            print(f"'{song}' not found in the playlist.")

    # Get method to access the playlist
    def get_songs(self):
        return self._songs.copy() 


playlist = Playlist()

playlist.add_song("Kesariya")
playlist.add_song("Believer")
playlist.add_song("Blinding Lights")

playlist.remove_song("Believer")

print("\nCurrent Playlist:")
print(playlist.get_songs())
```

```python
#20:4).Create an abstract class PaymentMethod with an abstract method pay(amount). Then, create two subclasses: UPI and CreditCard, each implementing the pay method with a print statement showing how the payment would be processed.<br><br><em><strong>Hint:</strong> Use the abc module for abstraction.</em>

from abc import ABC, abstractmethod

class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# Subclass for UPI payments
class UPI(PaymentMethod):

    def pay(self, amount):
        print(f"Processing UPI payment of ₹{amount}...")


# Subclass for Credit Card payments
class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Processing Credit Card payment of ₹{amount}...")

upi_payment = UPI()
card_payment = CreditCard()

upi_payment.pay(500)
card_payment.pay(1200)
```
