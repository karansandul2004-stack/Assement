```python
#Declare four variables in Python: one integer (number of followers), one float (average rating), one string (your favorite app's name), 
#and one boolean (is_premium_user). 
#Print each variable and its type using the type() function.

#3:1) variables

number_of_followers = 180000          #integer
average_rating = 3.5                  #float
favorite_app_name = "Instagram"       #string
is_premium_user = True                #boolean

# variable

print("Number_of_followers: ", number_of_followers)
print("average_rating: ", average_rating)
print("favorite_app_name:", favorite_app_name)
print("Is_premium_user: ", is_premium_user)

# datatype

print("\nData Type")
print("Number_of_followers: ", type(number_of_followers))
print("average_rating: ", type(average_rating))
print("favorite_app_name:", type(favorite_app_name))
print("Is_premium_user: ", type(is_premium_user))
```

```python
#Write a Python program that takes a user's input for the price of a Zomato order as a string, 
#converts it to a float using type casting, adds 18% GST, and prints the final bill amount.

#3:2) Zomato final billing with 18% GST

order_price = float(input("Enter the Zomato order price: "))

gst = order_price * 0.18

Final_bill = order_price + gst

print("Order Price :₹", order_price)
print("GST (18%): ₹", gst)
print("Final Bill Amount: ₹", Final_bill)
```

```python
#3:3)Given a list of strings representing product prices 
#from Flipkart, like ['199.99', '299.50', '150'], convert all to floats and calculate the total cart value.

prices = ['199.99','299.50','150']   #string

#string to float
price_float = [float(price) for price in prices]

total = sum(price_float)

#Display results
print("prices as float: ",price_float)
print("Total cart value: ₹", total)
```

```python
#3:4)Build a function is_discount_applicable(order_amount) that takes a float and returns True if the amount is greater than 500, otherwise False. 
#Print the result for order amounts 450 and 750.

def is_discount_application(order_amount):
    return order_amount > 500

print("Order amount: 450")
print("Discount Application:", is_discount_application(450))

print("Order amount: 750")
print("Discount Application:", is_discount_application(750))
```

```python
#3:5) You received a dataset of ratings as strings from Spotify: ['4.5', '3.0', '5', '4.2']. Use type casting to convert these to floats, then find and 
#print the highest rating.<br><br><em><strong>Hint:</strong> Use the float() function inside a loop or list comprehension.</em>

#List of strings
ratings = ['4.5','3.0','5','4.2']

#strings to floats
ratings_float = [float(rating) for rating in ratings]

#Find maximum rating
higest_rating = max(ratings_float)

print("Rating as floats:", ratings_float)
print("Higest rating:", higest_rating)
```
