```python
#4:1)Create a Python script that takes any product name string (e.g., 'Redmi Note 12 Pro') and prints the name in all uppercase 
#and all lowercase using the upper() and lower() methods.

product_name = "Redmi Note 12 Pro"

# uppercase and lowercase

print("Original Product:", product_name)
print("Uppercase:", product_name.upper())
print("Lowercase:", product_name.lower())
```

```python
#4:2). Write a function clean_brand_name(name) that removes leading/trailing spaces and replaces any 
#hyphens '-' with a single space in the input string. Test it with ' oneplus-Nord '.

def clean_brand_name(name):
    return name.strip().replace("-"," ")

brand = " oneplus-Nord "
cleaned_brand = clean_brand_name(brand)

print("Original:", repr(brand))
print("Cleaned:", cleaned_brand)
```

```python
# 4:3)Given the string 'Apple iPhone 14 Pro Max', use string slicing to extract and print only the brand name and the model 
#(i.e., 'Apple' and 'iPhone 14 Pro Max') separately.<br><br><em><strong>Hint:</strong> Use split() 
# to help find the split point, then use slicing for the substrings.</em>

#string
product = "Apple iPhone 14 Pro Max"

#first_space
split_index = product.find(" ")

#Extract brand name and model using slicing 
brand = product[:split_index]
model = product[split_index + 1:]

print("Brand:", brand)
print("Model:", model)
```

```python
#4:5)Suppose you have a list of messy product names: [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']. 
#Write code to clean each name (remove spaces, replace hyphens with spaces, and make the brand title case) and print the cleaned list.
#<br><br><em><strong>Constraint:</strong> Use at least three string methods from this session.</em>

# Answer
# List of messy product names
products = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']

# Clean product
cleaned_products = []

for product in products:
    cleaned = product.strip().replace("-", " ").title()
    cleaned_products.append(cleaned)

print("Cleaned Product Names:")
print(cleaned_products)
```
