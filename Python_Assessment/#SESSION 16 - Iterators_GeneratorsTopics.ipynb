```python
#16:2).Create a playlist of 6 songs (as a list of strings) and use enumerate() to print each song with its position like Spotify's tracklist (e.g., '1. Kesariya').

playlist = [
    "Kesariya",
    "Enemies",
    "Heeriye",
    "Mashooqa",
    "Mukabela",
    "Shape of You"
]

for position, song in enumerate(playlist, start=1):
    print(f"{position}. {song}")
```

```python
#16:3).Given two lists — one of food items and one of prices — use zip() to print each food item with its price like a Zomato menu (e.g., 'Pizza - ₹250').

food_items = ["Pizza", "Burger", "Pasta", "Bhel", "Sandwich"]

# prices
prices = [250, 180, 220, 250, 150]


print("----- Zomato Menu -----")
for item, price in zip(food_items, prices):
    print(f"{item} - ₹{price}")
```

```python
#16:4).Write a generator function called insta_posts_generator(posts) that takes a list of Instagram post captions and yields one caption at a time. Use next() to get and print the next post caption each time until all captions are printed.<br><br><em><strong>Hint:</strong> Use the yield keyword inside your function and handle StopIteration when all posts are done.</em>

def insta_posts_generator(posts):
    for post in posts:
        yield post


posts = [
    "Enjoying the weekend! 🌟",
    "Beautiful sunset today! 🌅",
    "Learning Python! 🐍",
    "Had a great day! 😊"
]

post_generator = insta_posts_generator(posts)

while True:
    try:
        print(next(post_generator))
    except StopIteration:
        print("All posts have been printed.")
        break
```

```python
#16:5).Build a generator function called cashback_generator(transactions) that takes a list of Paytm transaction amounts and yields 5% cashback for each transaction. Print out the cashback values for all transactions.

def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05  # 5% cashback


transactions = [500, 1200, 750, 300, 1000]

# the generator
cashback = cashback_generator(transactions)

# Cashback for each transaction
print("Paytm Cashback Details")
for cb in cashback:
    print(f"Cashback: ₹{cb:.2f}")
```
