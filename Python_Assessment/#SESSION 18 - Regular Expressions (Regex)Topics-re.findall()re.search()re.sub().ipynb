```python
#18:1).Use re.findall() to extract all valid Indian phone numbers (10 digits, starting with 7, 8, or 9) from a given text string that contains random numbers, prices, and phone numbers like those seen in OLX or WhatsApp chats.

import re

text = """
Contact me at 9876543210 or 8123456789.
Price: ₹2500, Order ID: 123456.
Old number: 6123456789
WhatsApp: 9988776655
Random digits: 1234567890
Another valid number: 7654321098
"""

pattern = r"\b[789]\d{9}\b"

# Extract all 
phone_numbers = re.findall(pattern, text)

print("Valid Indian Phone Numbers:")
for number in phone_numbers:
    print(number)
```

```python
#18:2).Write a Python function using re.search() that checks if a given string contains a valid date in the format DD/MM/YYYY (e.g., 25/06/2024), and returns True if found, otherwise False.<br><br><em><strong>Hint:</strong> Use the pattern '\b\d{2}/\d{2}/\d{4}\b'.</em>

import re

def contains_valid_date(text):
    # Expression for DD/MM/YYYY format
    pattern = r"\b\d{2}/\d{2}/\d{4}\b"

    # Search for the date pattern
    if re.search(pattern, text):
        return True
    return False


print(contains_valid_date("Meeting is on 25/06/2024."))   
print(contains_valid_date("Today's date is 2024-06-25.")) 
print(contains_valid_date("No date here."))               
```

```python
#18:3).Given a messy text copied from a Zomato review containing multiple emails, use re.findall() to extract all valid email addresses and print them as a list.

import re

# Messy text copied from a Zomato review
text = """
Amazing food! Contact us at support@zomato.com for help.
For business inquiries, email yakrut542@gmail.com.
My email is maitri.shah99@yahoo.co.in.
Random text: abc@, @gmail.com, user#mail.com
You can also reach us at feedback_2024@company.org.
"""

# Regular expression to match valid email addresses
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"


emails = re.findall(pattern, text)


print("Valid Email Addresses:")
print(emails)
```

```python
#18:4).Use re.sub() to mask all but the last 4 digits of any phone number in a string (e.g., replace 9876543210 with ******3210) like Paytm does for privacy.<br><br><em><strong>Constraint:</strong> Do not use loops; achieve this only with re.sub().</em>

import re

text = """
Call me at 9854499877 or 9115577644.
You can also reach me on 8824598763.
"""

# Mask all but the last 4 digits of 10-digit phone numbers
masked_text = re.sub(r"\b([7-9])\d{5}(\d{4})\b", r"******\2", text)

print(masked_text)
```

```python
#18:5).Use ChatGPT to generate a regex pattern that matches Flipkart-style order IDs (e.g., OD123456789012345000) and test it in Python using re.search() on sample order strings.

import re

# Regex pattern for Flipkart-style Order ID
pattern = r"\bOD\d{18}\b"

orders = [
    "Your Flipkart order ID is OD123456789012345000.",
    "Track your order: OD987654321098765432",
    "Invalid order: OD12345",
    "Order number: 123456789012345000",
    "No order ID available."
]

for order in orders:
    match = re.search(pattern, order)

    if match:
        print(f"Valid Order ID Found: {match.group()}")
    else:
        print("No valid Order ID found.")
```
