```python
#12:1). Write a lambda function that takes a price in rupees and returns the price after adding 18% GST. Test it on the prices 100, 250, and 500.

add_gst = lambda price: price * 1.18

prices = [100, 250, 500]

for price in prices:
    print(f"Original Price: ₹{price}, Price after 18% GST: ₹{add_gst(price):.2f}")
```

```python
#12:2). Given a list of song titles from Spotify with extra spaces and inconsistent casing, use map() and a lambda function to clean each title so that it is stripped of spaces and converted to title case (e.g., ' shape OF you ' → 'Shape Of You').

songs = [
    " shape OF you ",
    " believer ",
    " BLINDING lights ",
    " kesariya ",
    " peaches "
]

cleaned_songs = list(map(lambda song: song.strip().title(), songs))

print("Original Songs:")
print(songs)

print("\nCleaned Songs:")
print(cleaned_songs)
```

```python
#12:3). Use filter() and a lambda function to extract only those Flipkart product names from a list that start with the letter 'S' (case-insensitive).

products = [
    "Samsung Galaxy S24",
    "Redmi Note 13",
    "Sony Headphones",
    "Apple iPhone 15",
    "Smart Watch",
    "Boat Earbuds",
    "Sandisk SSD"
]

# Filter products that start with 'S' (case-insensitive)
s_products = list(filter(lambda product: product.lower().startswith("s"), products))

print("Products starting with 'S':")
print(s_products)
```

```python
#12:4). Given a list of order amounts from a Zomato cart [120, 340, 560, 80], use reduce() from functools to calculate the total bill amount.

from functools import reduce

order_amounts = [120, 340, 560, 80]

# Calculate the total bill amount using reduce()
total_bill = reduce(lambda x, y: x + y, order_amounts)

print("Order Amounts:", order_amounts)
print("Total Bill Amount: ₹", total_bill)
```

```python
#12:5). Use ChatGPT or Copilot to generate a Python code snippet that uses map(), filter(), and reduce() together to process a list of numbers: first double each number, then filter to keep only numbers greater than 100, and finally sum the result. Paste and test the generated code with the list [40, 60, 80, 120].

from functools import reduce

numbers = [40, 60, 80, 120]

# Double each number using map()
doubled = list(map(lambda x: x * 2, numbers))

# Keep only numbers greater than 100 using filter()
filtered = list(filter(lambda x: x > 100, doubled))

# Sum the remaining numbers using reduce()
total = reduce(lambda x, y: x + y, filtered)

print("Original List:", numbers)
print("Doubled List:", doubled)
print("Filtered List (>100):", filtered)
print("Sum of Filtered Numbers:", total)
```

```python
from functools import reduce

# Original list
numbers = [40, 60, 80, 120]

# Step 1: Double each number using map()
doubled = map(lambda x: x * 2, numbers)

# Step 2: Filter to keep only numbers greater than 100
filtered = filter(lambda x: x > 100, doubled)

# Step 3: Sum the result using reduce()
result = reduce(lambda a, b: a + b, filtered)

print("Final Result:", result)
```
