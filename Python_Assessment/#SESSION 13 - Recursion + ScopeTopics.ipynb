```python
#13:1). Write a recursive function in Python called reverse_string(s) that takes a string and returns it reversed (e.g., 'hello' becomes 'olleh').

def reverse_string(s):
    if len(s) <= 1:
        return s

    return reverse_string(s[1:]) + s[0]

text = "hello"
print("Original String:", text)
print("Reversed String:", reverse_string(text))
```

```python
#13:2). Build a recursive function sum_playlist_durations(durations) that takes a list of song durations (in seconds) and returns the total duration, similar to how Spotify totals a playlist.

def sum_playlist_durations(durations):
    
    if len(durations) == 0:
        return 0

    return durations[0] + sum_playlist_durations(durations[1:])

playlist = [210, 180, 240, 150, 300]  

total_duration = sum_playlist_durations(playlist)

print("Song Durations:", playlist)
print("Total Playlist Duration:", total_duration, "seconds")
```

```python
#13:4). Create a recursive function count_likes(posts) that takes a nested dictionary representing Instagram posts and their replies (each with a 'likes' key), and returns the total number of likes across all posts and replies.<br><br><em><strong>Hint:</strong> Each reply can itself have more replies, so use recursion to sum likes at all levels.</em>

def count_likes(posts):
    # Base case: if no posts, return 0
    if not posts:
        return 0
    
    total = 0
    for post in posts:
        # Add likes of the current post
        total += post.get('likes', 0)
        
        # Recursively add likes from replies (if any)
        if 'replies' in post:
            total += count_likes(post['replies'])
    
    return total


# Example usage:
posts_data = [
    {
        "likes": 10,
        "replies": [
            {"likes": 5},
            {"likes": 2, "replies": [{"likes": 3}]}
        ]
    },
    {
        "likes": 20,
        "replies": [
            {"likes": 7},
            {"likes": 4, "replies": [{"likes": 6}, {"likes": 1}]}
        ]
    }
]

print("Total Likes:", count_likes(posts_data))
```
