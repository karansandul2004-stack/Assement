```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Task 1
# 1. Create a sample Swiggy orders DataFrame with 4 payment types
np.random.seed(42)
data = {
    'Order_ID': [f'SW{i}' for i in range(1001, 1151)],
    'Payment_Method': np.random.choice(['UPI', 'Card', 'Cash', 'Net Banking'], size=150, p=[0.5, 0.3, 0.15, 0.05]),
    'Amount': np.random.randint(120, 800, size=150)
}
df = pd.DataFrame(data)

# 2. Set the visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 3. Create the countplot
# (Note: hue is mapped to x and legend=False is set to follow best practices in modern Seaborn)
sns.countplot(data=df, x='Payment_Method', hue='Payment_Method', palette='Set2', legend=False)

# 4. Add clear titles and labels
plt.title('Distribution of Swiggy Orders by Payment Method', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Payment Method', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# 5. Display the plot
plt.tight_layout()
plt.show()
```

```python
#Task 2
#Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) 
#from a Zomato-style dataset.
#<br><br><em><strong>Hint:</strong> Use the estimator parameter to show the mean delivery time for each category.</em>

# 1. Create a sample Zomato-style dataset
np.random.seed(42)
categories = ['Pizza', 'Burger', 'Biryani', 'South Indian', 'Desserts']

data = {
    'Food_Category': np.random.choice(categories, size=200),
    # Generate realistic delivery times between 15 and 60 minutes
    'Delivery_Time_Min': np.random.randint(15, 61, size=200) 
}
df = pd.DataFrame(data)

# 2. Set the visual theme
sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 5))

# 3. Create the barplot using the mean estimator
# By default, Seaborn calculates the mean (average), but we explicitly set estimator='mean'
# errorbar=None removes the confidence interval lines for a cleaner look
sns.barplot(data=df, x='Food_Category', y='Delivery_Time_Min', estimator='mean', errorbar=None,hue='Food_Category',palette='Spectral',legend=False)

# 4. Add titles and axis labels
plt.title('Average Delivery Time by Food Category (Zomato Dataset)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Food Category', fontsize=12)
plt.ylabel('Average Delivery Time (Minutes)', fontsize=12)

# 5. Display the plot
plt.tight_layout()
plt.show()
```

```python
# Task 3
#Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().
sns.barplot(data=df, x="Food_Category",y="Delivery_Time_Min",estimator="sum",errorbar=None,hue="Food_Category",palette="rocket",legend=False,)
```

```python
#Task 4
#Plot a countplot of movie genres from a BookMyShow-style dataset, but remove the confidence interval from the plot.
#<br><br><em><strong>Hint:</strong> Use the ci=None parameter to remove confidence intervals in sns.barplot().</em>

def generate_bookmyshow_data() -> pd.DataFrame:
    """Generates a sample BookMyShow-style dataset for movie genres."""
    np.random.seed(42)
    genres = ["Action", "Comedy", "Drama", "Thriller", "Sci-Fi"]
    data = {"movie_genre": np.random.choice(genres, size=300, p=[0.3, 0.25, 0.2, 0.15, 0.1])}
    return pd.DataFrame(data)


def plot_genre_counts(df: pd.DataFrame) -> None:
    """Plots a countplot of movie genres."""
    sns.set_theme(style="whitegrid")
    fig, ax = plt.subplots(figsize=(8, 5))

    # countplot automatically calculates counts and has no confidence intervals
    sns.countplot(data=df, x="movie_genre", hue="movie_genre", palette="Set2", legend=False,ax=ax)

    ax.set_title("Number of Movies by Genre (BookMyShow)", fontsize=14, fontweight="bold", pad=15)
    ax.set_xlabel("Movie Genre", fontsize=12)
    ax.set_ylabel("Movie Count", fontsize=12)

    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    movie_df = generate_bookmyshow_data()
    plot_genre_counts(movie_df)
```
