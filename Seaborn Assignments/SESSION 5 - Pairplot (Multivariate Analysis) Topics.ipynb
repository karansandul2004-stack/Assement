```python
import seaborn as sns
sns.get_dataset_names()
```

```python
import matplotlib.pyplot as plt
```

```python
#1.Load the 'tips' dataset from seaborn and create a basic pairplot for all numerical columns.

#Set the visual theme for clean looking grids
sns.set_theme(style="ticks")
#Load the built-in 'tips' dataset from Seaborn
# This dataset contains information about restaurant bills, tips given, gender, day, etc.
tips_df = sns.load_dataset("tips")

sns.pairplot(data = tips_df)

plt.show()
```

```python
#2.Use the pairplot function to visualize the 'tips' dataset, 
#but this time set the hue parameter to 'sex' and observe how the data points are colored based on gender.

sns.pairplot(data = tips_df, hue = "sex", palette = "Set1")

plt.show()
```

```python
#3.Create a pairplot for the 'tips' dataset with the hue set to 'smoker', and set the diag_kind 
#parameter to 'kde' to show kernel density estimates on the diagonal plots.<br><br><em><strong>Hint:</strong> Use diag_kind='kde' 
#in the pairplot function.</em>

#Create the pairplot with hue='smoker' and diag_kind='kde'
sns.pairplot(data = tips_df, hue = "smoker", diag_kind = "kde", palette = "Dark2")

plt.show()
```

```python
#4.Pick any two numerical columns from the 'tips' dataset and 
#use pairplot to visualize only those columns, setting hue to 'time' (Lunch/Dinner).<br><br><em><strong>Constraint:</strong> 
#Use the vars parameter to select the columns.</em>

#We pick 'total_bill' and 'tip' as our two columns, and group by dining time

sns.pairplot(data = tips_df, vars = ["total_bill", "tip"], hue = "time", palette = "Accent")

plt.show()
```
