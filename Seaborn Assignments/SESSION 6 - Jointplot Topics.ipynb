```python
#1.Load a CSV 
#file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to visualize the relationship 
#between rating and price.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Create a sample CSV file so the code runs directly

np.random.seed(42)
sample_data ={
    'price_inr' : np.random.randint(499, 4500, size = 150),
    'product_rating' : np.random.uniform(2.5, 5.0, size = 150).round(1)
}

pd.DataFrame(sample_data).to_csv('flipkart_data.csv', index = False)

# Load Csv File

df_flipkart = pd.read_csv('flipkart_data.csv')

# Set the visual theme
sns.set_theme(style="darkgrid")

# Create the jointplot with kind='scatter'
# This maps 'product_rating' on the x-axis and 'price_inr' on the y-axis

g = sns.jointplot(data=df_flipkart, x='product_rating',y='price_inr',kind='scatter',color='purple',alpha=0.7)

#Add a clean title using the jointplot grid structure

g.fig.suptitle('Relationship Between Flipkart Product Ratings & Prices', fontsize=14, fontweight='bold', y=1.02)
g.set_axis_labels('Product Rating (out of 5)', 'Price (in INR)', fontsize=12)


plt.show()
```

```python
#2.Create a jointplot with kind='hex' to analyze the relationship between delivery time and order value using a dataset of Swiggy food orders.<br><br>

#Create a sample dataset of 500 Swiggy orders
np.random.seed(42)
swiggy_data = {
    # Most orders take between 20 to 50 minutes
    "delivery_time_min": np.random.normal(loc=35, scale=8, size=500).astype(int),
    # Most orders cost between ₹150 and ₹800
    "order_value_inr": np.random.randint(150, 800, size=500),
}
df_swiggy = pd.DataFrame(swiggy_data)

# Set the visual theme for clean looking grids
sns.set_theme(style="ticks")

# Create the jointplot with kind='hex'
# gridsize controls how large or small the hexagons look
g = sns.jointplot(data=df_swiggy,x="delivery_time_min",y="order_value_inr",kind="hex",color="orangered",gridsize=15)

#Add titles and clean up labels
g.fig.suptitle("Swiggy Orders: Delivery Time vs. Order Value",fontsize=14,fontweight="bold",y=1.02)
g.set_axis_labels("Delivery Time (Minutes)", "Order Value (INR)", fontsize=12)

plt.show()
```

```python
#3.Use Seaborn's jointplot with kind='kde' to visualize how the number of likes and comments are related in a dataset of Instagram posts.

#Create a sample dataset of 200 Instagram posts
np.random.seed(8)
likes = np.random.randint(50, 2000, size=200)
# Generally, more likes mean more comments, so we link them together mathematically
comments = (likes * np.random.uniform(0.02, 0.08, size=200)).astype(int) + np.random.randint(2, 15, size=200)

instagram_data = {
    'likes_count': likes,
    'comments_count': comments
}
df_instagram = pd.DataFrame(instagram_data)

# Set the visual theme for clean looking grids
sns.set_theme(style="white")

# Create the jointplot with kind='kde'
# fill=True shades the spaces between the rings for an engaging look
g = sns.jointplot(data=df_instagram,x='likes_count',y='comments_count',kind='kde',color='magenta',fill=True,thresh=0.05)

# Add titles and clean up labels
g.fig.suptitle('Instagram Engagement: Likes vs. Comments Density', fontsize=14, fontweight='bold', y=1.02)
g.set_axis_labels('Number of Likes', 'Number of Comments', fontsize=12)

plt.show()
```

```python
#4.Compare the visual insights you get from scatter, hexbin, 
#and KDE jointplots using the same dataset of Zomato restaurant reviews (rating vs number of reviews). 
#Briefly describe which plot type helps you understand the data best and why.

# Create a sample Zomato dataset (500 restaurants)
np.random.seed(42)
ratings = np.random.uniform(2.0, 5.0, size=500)
# Popular restaurants tend to have higher ratings and way more reviews
reviews = (np.exp(ratings) * np.random.uniform(5, 30, size=500)).astype(int)

zomato_data = {"restaurant_rating": ratings, "num_of_reviews": reviews}
df_zomato = pd.DataFrame(zomato_data)

# Set the visual theme
sns.set_theme(style="ticks")

# PLOT 1: Scatter Jointplot
g1 = sns.jointplot(data=df_zomato,x="restaurant_rating",y="num_of_reviews",kind="scatter",color="crimson",alpha=0.6)
g1.fig.suptitle("1. Scatter Plot View", fontsize=12, fontweight="bold", y=1.02)
g1.set_axis_labels("Restaurant Rating", "Number of Reviews")

# PLOT 2: Hexbin Jointplot
g2 = sns.jointplot(data=df_zomato,x="restaurant_rating",y="num_of_reviews",kind="hex",color="darkred",gridsize=15)
g2.fig.suptitle("2. Hexbin Plot View", fontsize=12, fontweight="bold", y=1.02)
g2.set_axis_labels("Restaurant Rating", "Number of Reviews")

# PLOT 3: KDE Jointplot
g3 = sns.jointplot(data=df_zomato,x="restaurant_rating",y="num_of_reviews",kind="kde",color="red",fill=True,thresh=0.05)
g3.fig.suptitle("3. KDE Contour View", fontsize=12, fontweight="bold", y=1.02)
g3.set_axis_labels("Restaurant Rating", "Number of Reviews")


plt.show()
```
