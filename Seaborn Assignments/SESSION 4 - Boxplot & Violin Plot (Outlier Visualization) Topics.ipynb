```python
#1.Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample CSV file so the code runs directly
sample_data = {
    'product_name': [f'Product_{i}' for i in range(100)],
    'price_inr': np.random.randint(299, 2499, size=100)
}
pd.DataFrame(sample_data).to_csv('flipkart_prices.csv', index=False)


# Load the CSV file
df_flipkart = pd.read_csv('flipkart_prices.csv')

# Set up the window
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 4))

# Draw the boxplot (horizontal view using x)
sns.boxplot(data=df_flipkart, x='price_inr', color='skyblue')

# Add student-friendly labels
plt.title('Distribution of Flipkart Product Prices', fontsize=14, fontweight='bold')
plt.xlabel('Price (in INR)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#2.Use sns.violinplot() to compare the order amounts for 
#two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset.

# Create sample data for Fast Food and Desserts
np.random.seed(15)
categories = ['Fast Food'] * 50 + ['Desserts'] * 50

# Fast Food typically costs a bit more per order than standard desserts
amounts = list(np.random.randint(150, 600, size=50)) + list(np.random.randint(80, 350, size=50))

zomato_data = {
    'restaurant_category': categories,
    'order_amount_inr': amounts
}
df_zomato = pd.DataFrame(zomato_data)

# Set up the window
plt.figure(figsize=(8, 5))

# Draw the violin plot
sns.violinplot(data=df_zomato, x='restaurant_category', y='order_amount_inr', hue='restaurant_category', palette='Set2',legend=False)

# Add labels
plt.title('Zomato Order Amounts: Fast Food vs. Desserts', fontsize=14, fontweight='bold')
plt.xlabel('Restaurant Category', fontsize=12)
plt.ylabel('Order Amount (INR)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#3.Given an array of IPL match ticket prices, plot a boxplot using sns.boxplot()
#and identify any outliers in the data.<br><br><em><strong>Hint:</strong> Outliers are points that appear outside the whiskers of the boxplot.</em>

# Create an array of normal ticket prices, plus a few extreme VIP ticket prices (outliers)
np.random.seed(5)
normal_tickets = np.random.randint(800, 2500, size=40)
vip_tickets = np.array([6000, 7500, 8000]) # These will be our clear outliers!

ticket_prices = np.concatenate([normal_tickets, vip_tickets])
df_ipl = pd.DataFrame({'ticket_price_inr': ticket_prices})

# Set up the window
plt.figure(figsize=(9, 4))

# Draw the boxplot
sns.boxplot(data=df_ipl, x='ticket_price_inr', color='lightcoral')

# Add labels
plt.title('IPL Match Ticket Prices & Outlier Detection', fontsize=14, fontweight='bold')
plt.xlabel('Ticket Price (INR)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#4.Create a violin plot using sns.violinplot() to show 
#the distribution of daily step counts collected from a fitness app like Google Fit, and explain what the shape of the plot tells 
#you about user activity.

#Create a month of step data (mostly regular days, but active weekends)
np.random.seed(3)
daily_steps = np.random.normal(loc=7000, scale=1200, size=30).astype(int)
df_steps = pd.DataFrame({'google_fit_steps': daily_steps})

# Set up the window
plt.figure(figsize=(6, 5))

# Draw the vertical violin plot
sns.violinplot(data=df_steps, y='google_fit_steps', color='mediumaquamarine', inner='box')

# Add labels
plt.title('Distribution of Daily Step Counts (Google Fit)', fontsize=14, fontweight='bold')
plt.ylabel('Daily Step Count', fontsize=12)

plt.tight_layout()
plt.show()
```
