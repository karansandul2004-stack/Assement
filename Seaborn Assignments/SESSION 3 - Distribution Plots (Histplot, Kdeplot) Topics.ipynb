```python
#1.Import Seaborn and 
#use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Create sample data (50 Swiggy delivery times around 35 minutes)
np.random.seed(42)
delivery_data = {
    'delivery_time_min': np.random.normal(loc=35, scale=8, size=50).astype(int)
}
df_swiggy = pd.DataFrame(delivery_data)

#Set up the plotting window style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

#Draw the histogram
sns.histplot(data=df_swiggy, x='delivery_time_min', color="coral", edgecolor="black")

#Add student-friendly labels
plt.title('Distribution of Swiggy Delivery Times', fontsize=14, fontweight='bold')
plt.xlabel('Delivery Time (in Minutes)', fontsize=12)
plt.ylabel('Number of Orders (Count)', fontsize=12)

#Show the window
plt.tight_layout()
plt.show()
```

```python
#2.Download or create a list of daily step counts (e.g., from a fitness app) for one month and 
#plot the distribution using sns.kdeplot(), showing the probability density curve.

#Create sample data (30 days of step counts averaging 8500 steps)
np.random.seed(24)
fitness_data = {
    'daily_steps': np.random.normal(loc=8500, scale=1500, size=30).astype(int)
}
df_fitness = pd.DataFrame(fitness_data)

#Set up the window
plt.figure(figsize=(8, 5))

#Draw the KDE curve
#fill=True shades the area under the curve; alpha sets transparency (0.3 = 30% solid)
sns.kdeplot(data=df_fitness, x='daily_steps', color="teal", fill=True, alpha=0.3)

#Labels
plt.title('Probability Density of Daily Step Counts (1 Month)', fontsize=14, fontweight='bold')
plt.xlabel('Daily Steps Taken', fontsize=12)
plt.ylabel('Density (Likelihood)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#3.Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, 
#using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:
#</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

# Create sample data (100 Zomato monthly bills in Rupees)
np.random.seed(10)
zomato_data = {
    'monthly_spending_inr': np.random.gamma(shape=5, scale=120, size=100).astype(int)
}
df_zomato = pd.DataFrame(zomato_data)

# Set up the window
plt.figure(figsize=(8, 5))

# Draw both together using the built-in kde=True shortcut parameter
sns.histplot(data=df_zomato, x='monthly_spending_inr', kde=True, color="crimson", edgecolor="black")

# Labels
plt.title('Zomato Monthly Spending Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Spending Amount (INR)', fontsize=12)
plt.ylabel('Count / Density Line', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#4.Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. 
#Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5, bins=10, and bins=20.</em>

# Step 1: Create sample data (150 simulated IPL innings scores)
np.random.seed(7)
ipl_data = {
    'innings_score': np.random.randint(120, 241, size=150)
}
df_ipl = pd.DataFrame(ipl_data)

# --- GRAPH A: Broad view (bins=5) ---
plt.figure(figsize=(6, 4))
sns.histplot(data=df_ipl, x='innings_score', bins=5, color="#4C72B0", edgecolor="black")
plt.title("IPL Scores (bins=5) - Too Chunky")
plt.show()

# --- GRAPH B: Balanced view (bins=10) ---
plt.figure(figsize=(6, 4))
sns.histplot(data=df_ipl, x='innings_score', bins=10, color="#DD8452", edgecolor="black")
plt.title("IPL Scores (bins=10) - Just Right")
plt.show()

# --- GRAPH C: Detailed/Noisy view (bins=20) ---
plt.figure(figsize=(6, 4))
sns.histplot(data=df_ipl, x='innings_score', bins=20, color="#55A868", edgecolor="black")
plt.title("IPL Scores (bins=20) - Jagged / Noisy")
plt.show()
```
