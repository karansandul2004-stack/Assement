```python
#Task 2
#Using Seaborn's lineplot(), plot the daily temperature trend for one week using the following data: days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
#temps = [32, 34, 33, 35, 36, 37, 34]. Label the axes and title the chart 'Weekly Temperature Trend'.

days = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
temps = [32,34,33,35,36,37,34]

sns.lineplot(x= days, y= temps)

plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# Task 3
#Recreate the above line plot using Seaborn's axes-level and figure-level functions separately. Briefly note one difference 
#you observe in how these functions work.<br><br><em><strong>Hint:</strong> Use sns.lineplot() for axes-level and sns.relplot() for figure-level.</em>

ax = sns.lineplot(x=days,y=temps)

# 1. Axes-level approach
ax.set_title("Weekly Temperature Trend (axes.level)")

plt.show()

```

```python
# 2. Figure-level function
g = sns.relplot(x=days,y=temps,kind = "line")

g.figure.suptitle("Weekly Temperature Trend (Figure-Level)", y = 1.03)

plt.show()
```

```python
# Task 4
#Apply each of Seaborn's built-in themes (darkgrid, whitegrid, ticks) to your temperature trend plot and save three separate images. Compare 
#how the grid and background change with each theme.<br><br><em><strong>Hint:</strong> Use sns.set_theme(style='theme_name') before plotting.</em>

# 1. Darkgrid Theme
sns.set_theme(style="darkgrid")
plt.figure()
sns.lineplot(x=days, y=temps)
plt.title("Weekly Temperature Trend (Darkgrid)")
plt.savefig("temp_trend_darkgrid.png", bbox_inches="tight")
plt.close()
```

```python
# 2. Whitegrid Theme
sns.set_theme(style="whitegrid")
plt.figure()
sns.lineplot(x=days, y=temps)
plt.title("Weekly Temperature Trend (Whitegrid)")
plt.savefig("temp_trend_whitegrid.png", bbox_inches="tight")
plt.close()
```

```python
# 3. Ticks Theme
sns.set_theme(style="ticks")
plt.figure()
sns.lineplot(x=days, y=temps)
plt.title("Weekly Temperature Trend (Ticks)")
plt.savefig("temp_trend_ticks.png", bbox_inches="tight")
plt.close()
```
